## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

### Leakage Audit

I reviewed the features used in my Week 5 model to check for possible data leakage.

| **Feature** | **Leakage Risk** | **Reason** |
|---|---|---|
| Clicks | Low | Represents historical performance and is available before prediction. |
| Impressions | Low | Historical search visibility, not future information. |
| CTR | Medium | Derived from clicks and impressions, so it should be checked carefully. |
| Average Position | Low | Search ranking information available at prediction time. |

### Conclusion

I did not find clear evidence of target leakage in my final feature set. However, derived features such as CTR should always be reviewed carefully to ensure they do not indirectly reveal the target.

```python
print("Leakage Audit")

features = {
    "Clicks": "Low Risk",
    "Impressions": "Low Risk",
    "CTR": "Medium Risk (derived feature)",
    "Average Position": "Low Risk"
}

for feature, risk in features.items():
    print(f"{feature}: {risk}")

print("\nConclusion:")
print("No obvious target leakage was observed in the selected features.")
```
